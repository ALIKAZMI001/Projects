```python
from transformers import pipeline

# Load sentiment analysis pipeline (uses DistilBERT by default)
classifier = pipeline("sentiment-analysis")

# Run prediction
result = classifier("I absolutely love this movie!")
print(result)
```

```python
from transformers import pipeline, set_seed

# Create text generation pipeline using GPT-2
generator = pipeline("text-generation", model="gpt2")
set_seed(42)

# Basic Q&A prompt
prompt = "Q: Who is the founder of OpenAI?\nA:"
output = generator(prompt, max_length=50, num_return_sequences=1)

print(output[0]['generated_text'])
```

```python
from transformers import pipeline

# T5 and Flan-T5 are designed for question-answering tasks
qa = pipeline("text2text-generation", model="google/flan-t5-base")

# Prompt with context
context = """OpenAI is an AI research lab founded in 2015. Its founders include Sam Altman and Elon Musk."""
question = f"Answer the question based on the context:\nContext: {context}\nQuestion: Who founded OpenAI?"

output = qa(question)
print(output[0]['generated_text'])

```

```python


# STEP 2: Import libraries
from transformers import pipeline
import fitz  # PyMuPDF
from google.colab import files

# STEP 3: Upload the PDF file
print("🔼 Please upload your PDF file (e.g., Cricket_Introduction.pdf)")
uploaded = files.upload()

# STEP 4: Get uploaded file name
pdf_path = list(uploaded.keys())[0]  # Gets the uploaded filename

# STEP 5: Extract text from the PDF
doc = fitz.open(pdf_path)
context = ""
for page in doc:
    context += page.get_text()
doc.close()

print("✅ PDF text extracted!")

# STEP 6: Load Flan-T5 model
qa = pipeline("text2text-generation", model="google/flan-t5-base")

# STEP 7: Ask your question
question = "What are different formats of cricket?"

# STEP 8: Build the prompt
prompt = f"Answer the question based on the context:\nContext: {context}\nQuestion: {question}"

# STEP 9: Generate the answer
output = qa(prompt, max_length=100)
print("🧠 Answer:", output[0]['generated_text'])
```

```python
from transformers import pipeline

# Step 3: Load the translation model
translator = pipeline("translation_en_to_ur", model="Helsinki-NLP/opus-mt-en-ur")

# Step 4: Translate your text
text = "The batsman hit the ball out of the stadium."
translation = translator(text, max_length=100)

# Step 5: Print the translated result
print("Urdu Translation:", translation[0]['translation_text'])
```
